# Regex

Read `FITZ-TGG-25_chapt1.txt`

```python
with open('../data/FITZ-TGG-25_chapt1.txt', 'r') as file:
    text = file.read()
```

Check that the script can read the text, by printing the first 100 characters.

```python
print(text[:100])
```

```python
import re
dialogues = re.findall(r'“.*?”', text)
for dialogue in dialogues:
    print(dialogue)
    print() 
```

Calculate the percentage of dialogue in the text.

```python
total_characters = len(text)
total_dialogue_characters = sum(len(dialogue) for dialogue in dialogues)
print("Number of characters in the text:", total_characters)
print("Number of characters in the dialogues:", total_dialogue_characters)
print("Percentage of dialogue in the text:", (total_dialogue_characters / total_characters) * 100)
```
