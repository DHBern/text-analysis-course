```python
sample_text = """In my younger and more vulnerable years my father gave me some advice
that I've been turning over in my mind ever since.
"Whenever you feel like criticizing anyone," he told me, "just
remember that all the people in this world haven't had the advantages
that you've had." """
```

## Tokenisation

See https://spacy.io/usage/linguistic-features#tokenization

> During processing, spaCy first tokenizes the text, i.e. segments it into words, punctuation and so on. This is done by applying rules specific to each language. For example, punctuation at the end of a sentence should be split off – whereas “U.K.” should remain one token.

> First, the raw text is split on whitespace characters[...]. Then, the tokenizer processes the text from left to right. On each substring, it performs two checks: 1) Does the substring match a tokenizer exception rule? For example, “don’t” does not contain whitespace, but should be split into two tokens, “do” and “n’t”, while “U.K.” should always remain one token. 2) Can a prefix, suffix or infix be split off? For example punctuation like commas, periods, hyphens or quotes.

Before using `spaCy`, let's try to tokenise by simply **splitting a text where a whitespace occurs (method `split()`)**.

We will print the list of tokens and the number of tokens.

```python
sample_tokens = sample_text.split()
print(sample_tokens)
print("Token count:", len(sample_tokens))
```

Now let's **tokenise using the function we used in session 3**.

```python
import re

def tokenize(text):
    # Brief description of the function
    """Return lowercase word tokens, keeping internal apostrophes."""

    # Finds all matches of one or more letters
    # [A-Za-z]+
    # optionally followed by an apostrophe (' or curly ’) and more letters
    # (?:['’][A-Za-z]+)?
    # 
    # Normalize case so capitalization does not create extra word types.
    # 
    # Return every match in a list of strings
    return re.findall(r"[A-Za-z]+(?:['’][A-Za-z]+)?", text.lower())

sample_tokens = tokenize(sample_text)
print("Tokens:", sample_tokens)
print("Token count:", len(sample_tokens))
```

Now let's **tokenise with `spaCy`**.

```python
import spacy

nlp = spacy.load("en_core_web_sm")
doc = nlp(sample_text)
for token in doc:
    print(token.text)
print("Token count:", len(doc))
```

Observe the difference in the output.

## POS tagging and lemmatisation

See https://spacy.io/usage/linguistic-features#pos-tagging.

> After tokenization, spaCy can parse and tag a given `Doc`. This is where the trained pipeline and its statistical models come in, which enable spaCy to make predictions of which tag or label most likely applies in this context. A trained component includes binary data that is produced by showing a system enough examples for it to make predictions that generalize across the language – for example, a word following “the” in English is most likely a noun.

## List of lemmas in order of frequency


```python
## Count and display lemmas in the document

from collections import Counter  # import Counter to count occurrences of each lemma
lemmas_in_order = [token.lemma_ for token in doc]  # create a list of lemmas from the document
lemmas_count = Counter(lemmas_in_order)  # count the occurrences of each lemma
lemmas_in_order_by_count = [lemma for lemma, count in lemmas_count.most_common()]  # list lemmas in descending order of count
from tabulate import tabulate  # import tabulate to print a nice table
print(tabulate(lemmas_count.most_common(), headers=["Lemma", "Count"], tablefmt="grid"))  # print the table of lemmas and their counts
```

## List of POS in order of frequency

```python
# Display the POS (part-of-speech) tags in the document in a table sorted by their frequency
from collections import Counter  # import Counter to count occurrences of each POS tag
pos_in_order = [token.pos_ for token in doc]  # create a list of POS tags from the document
pos_count = Counter(pos_in_order)  # count the occurrences of each POS tag
from tabulate import tabulate  # import tabulate to print a nice table
print(tabulate(pos_count.most_common(), headers=["POS", "Count"], tablefmt="grid"))  # print the table of POS tags and their counts
```

## Stop words

For each token, `spaCy` does not only recognise the POS and lemma, but also other features, including if the token is a stop word or not.

Let's recount the frequency of lemmas and POS tags, without the stopwords.


```python
## Recount the frequency of lemmas and POS tags, excluding stopwords
lemmas_no_stopwords = [token.lemma_ for token in doc if not token.is_stop]  # create a list of lemmas excluding stopwords
lemmas_no_stopwords_count = Counter(lemmas_no_stopwords)  # count the occurrences of each lemma excluding stopwords
print(tabulate(lemmas_no_stopwords_count.most_common(), headers=["Lemma", "Count"], tablefmt="grid"))  # print the table of lemmas and their counts excluding stopwords

pos_no_stopwords = [token.pos_ for token in doc if not token.is_stop]  # create a list of POS tags excluding stopwords
pos_no_stopwords_count = Counter(pos_no_stopwords)  # count the occurrences of each POS tag excluding stopwords
print(tabulate(pos_no_stopwords_count.most_common(), headers=["POS", "Count"], tablefmt="grid"))  # print the table of POS tags and their counts excluding stopwords    

```

## All together

```python
# Functions to get lemmas and POS excluding stopwords based on each token's is_stop attribute

def get_lemmas_pos_no_stopwords(doc, stopwords=True):
    # Create a list of tuples (lemma, POS) for each token in the doc
    # Exclude tokens that are stopwords if stopwords is True
    lemmas_pos_no_stopwords = [
        (token.lemma_, token.pos_)
        for token in doc
        if not stopwords or not token.is_stop
    ]
    # Return a Counter object with the frequency of each (lemma, POS) tuple
    return Counter(lemmas_pos_no_stopwords)

def get_pos_lemmas_no_stopwords(doc, stopwords=True):
    # Create a list of tuples (POS, lemma) for each token in the doc
    # Exclude tokens that are stopwords if stopwords is True
    pos_lemmas_no_stopwords = [
        (token.pos_, token.lemma_)
        for token in doc
        if not stopwords or not token.is_stop
    ]
    # Return a Counter object with the frequency of each (POS, lemma) tuple     
    return Counter(pos_lemmas_no_stopwords)
```

```python
# Run the functions to get lemmas and POS excluding stopwords
lemmas_pos_no_stopwords_count = get_lemmas_pos_no_stopwords(doc, stopwords=True)
print(tabulate(lemmas_pos_no_stopwords_count.most_common(), headers=["Lemma", "POS", "Count"], tablefmt="grid"))

pos_lemmas_no_stopwords_count = get_pos_lemmas_no_stopwords(doc, stopwords=True)
print(tabulate(pos_lemmas_no_stopwords_count.most_common(), headers=["POS", "Lemma", "Count"], tablefmt="grid"))

# Run the functions to get lemmas and POS with stopwords
lemmas_pos_with_stopwords_count = get_lemmas_pos_no_stopwords(doc, stopwords=False)
print(tabulate(lemmas_pos_with_stopwords_count.most_common(), headers=["Lemma", "POS", "Count"], tablefmt="grid"))

pos_lemmas_with_stopwords_count = get_pos_lemmas_no_stopwords(doc, stopwords=False)
print(tabulate(pos_lemmas_with_stopwords_count.most_common(), headers=["POS", "Lemma", "Count"], tablefmt="grid"))
```

## Save to a file

```python
# Define a function to save each table to a file inside a folder 'output'.
import os
import csv

def save_counts_to_csv(counts, filename):
    os.makedirs("output", exist_ok=True)
    outfile = f"output/{filename}.csv"
    with open(outfile, "w", newline="") as csvfile:
        writer = csv.writer(csvfile)
        for key, count in counts.most_common():
            writer.writerow(list(key) + [count])
```

```python
# Run the function to save the counts to a CSV file
save_counts_to_csv(lemmas_pos_no_stopwords_count, "lemmas_pos_no_stopwords_sample-text")
```
